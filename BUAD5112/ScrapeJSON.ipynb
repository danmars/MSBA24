## Obtaining Data

```python
# Import libraries.
import requests
import pandas as pd
```

```python
## Obtain the JSON data from the official portal for European data.
# URL was obtained by inspecting 'https://data.europa.eu/catalogue-statistics/evolution/?locale=en'.
search_url = 'https://data.europa.eu/api/hub/statistics/data/num-datasets'
header = {'User-Agent': 'Mozilla/5.0'}
response = requests.get(search_url, headers=header)

root=response.json()

print(root)
```

## Programming

```python
## Create the code in Python to obtain the data, and get it into a table using pandas.
df = pd.DataFrame(root['stats'])
```
