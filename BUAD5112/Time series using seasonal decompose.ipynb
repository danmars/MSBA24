We can do a time series decomposition, using the moving average method, relatively easy with the seasonal_decompose function of the statsmodel library.

```python
import pandas as pd
import numpy as np
from statsmodels.tsa.seasonal import seasonal_decompose

df=pd.read_csv('SSFb.csv')
 
df.head()
```

```python
df.set_index('Month',inplace=True)
df.index=pd.to_datetime(df.index)
#drop null values
df.dropna(inplace=True)
df.plot()
```

If the trend and seasonality look relatively stable, we can use the additive model. Otherwise, we are better off using the multiplicative model.

```python
result=seasonal_decompose(df['Sales'], model='additive', period=12)
```

```python
result.seasonal.plot()
```

```python
result.trend.plot()
```

```python
result.plot()
```
