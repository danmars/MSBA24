Importing the necessary libraries

```python
import requests  # Internet information requests
from bs4 import BeautifulSoup  # Parsing HTML
import pandas as pd # Enables code to create a sublist using a dataframe 
```

Getting and parsing the data

```python

url = 'https://publicinterestlegal.org/county-list/'

html_doc = requests.get(url).content

#parses html 
parsed_html = BeautifulSoup(html_doc)



new = parsed_html.find_all('tr')

print (new)
```

Creating the list of sublists

```python
list = []
county_names = []
states = []
registration_rates = []
for row in new:
    info = row.text.split('\n')[1:4]
    list.append(info)

headings = list[0]
county_info = list[1:]


for county in county_info:
    county_names.append(county[0])
    states.append(county[1])
    registration_rates.append(county[2])

print(list)
```

Getting the list into a data frame

```python

data = {headings[0]: county_names, headings[1]: states, headings[2]: registration_rates}
df = pd.DataFrame(data=data)

print("Result Length: ",(len(df)))
print()
df
```
