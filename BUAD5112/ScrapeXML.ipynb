## Obtaining Data

```python
# Import libraries.
import requests
from lxml import objectify

# Send GET request to specified URL.
URL = 'https://data.norfolk.gov/api/views/u85v-sad9/rows.xml?accessType=DOWNLOAD'
response = requests.get(URL).content

# Parse XML file.
root = objectify.fromstring(response)

# Reivew data.
print(response)
```

## Programming

### Task 1: Print the name of the plant delivered for site # 3F at 3031 Luxemburg Ave on 1/24/2018

```python
## Print the name of the plant delivered for site # 3F at 3031 Luxemburg Ave on 1/24/2018.
# Find rows with "3F" as the site, "3031" as the street_number, "Luxemburg Ave" as the street_name, and "2018-01-24" contained in delivered.
rows = root.xpath('//row/row[site = "3F" and street_number = "3031" and street_name = "Luxemburg Ave" and contains(delivered, "2018-01-24")]')

# Iterate through rows and print common name for each row.
for row in rows:
    print('Common Name:', row.common_name.text)
```

```python
# This task can also be completed by referencing the specific element/row, if known.
print('Common Name:',root.row.row[2].common_name)
```

###  Task 2: Print the source of the Two-winged Silverbell for 8075 West Glenn Rd

```python
## Print the source of the Two-winged Silverbell for 8075 West Glenn Rd.
# Find rows with "Two-winged Silverbell" as the common_name, "8075" as the street_number, and "West Glen Rd" as the street_name.
rows = root.xpath('//row/row[common_name = "Two-winged silverbell" and street_number = "8075" and street_name = "West Glen Rd"]')

# Iterate through rows and add sources to a set. By storing these elements to a set, we remove possible duplicates.
sources = set()
for row in rows:
    sources.add(row.source.text)

# Print unique sources.
for source in sources:
    print('Source:',source)
```

```python
# If we are not concerned about duplicates, then we simply interate through rows and print the source.
# Three Two-winged Silverbell trees have been delivered to 8075 West Glenn Rd, all sourced from the City Nursery.
for row in rows:
    print('Source:',row.source.text)
```

### Task 3: List the instances where Redbud has been delivered, including the date of delivery and address.

```python
## List the instances where Redbud has been delivered, including the date of delivery and address.
# Find rows with "Redbud" as common_name.
rows = root.xpath('//row/row[common_name = "Redbud"]')

# Iterate through rows and print delivery date and address.
print('Redbud Deliveries')
for row in rows:
    print('Delivery Date:', row.delivered.text, 'Address:',row.street_number.text, row.street_name.text)
```
