We will scrape information contained in the National Center for Environmental Information (NOAA). In particular, we are interested in finding out about temperature values and rankings for the different states in different periods of time.

Go to https://www.ncei.noaa.gov/access/monitoring/climate-at-a-glance/statewide/rankings

Let's explore the average temperatures for November of 2021 in Virginia.

```python
import requests
from lxml import objectify


URL = 'https://www.ncei.noaa.gov/access/monitoring/climate-at-a-glance/statewide/rankings/44-tavg-202111/data.xml'
response = requests.get(URL).content

import requests
from lxml import objectify

#parsing the XML file
root = objectify.fromstring(response)

#take a look at the data
print(response)
```

Based on the structure of the file, let's get the values of several indicators for the month.

```python
print (root.description[0].title)


print('Average temperature rankings for period ', root.data[0].period)
print('Average for the period: ', root.data[0].value)
print('Mean value 1901-2000: ',root.data[0].mean)
print('Anomaly: ',root.data[0].departure)
print('Low rank: ',root.data[0].lowRank)
print('High rank: ',root.data[0].highRank)
```

Now let's do the same, but for the period between July and November of the same year.
