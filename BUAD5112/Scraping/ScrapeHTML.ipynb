## Obtaining Data

```python
# Import libraries.
import requests  # Internet information requests
from bs4 import BeautifulSoup  # Parsing HTML
import pandas as pd # Enables code to create a sublist using a dataframe
```

```python
## Write the code in Python to obtain and parse the data with requests and BeautifulSoup
# Send GET request to specified URL.
url = 'https://gpa.eng.uci.edu/'
html_doc = requests.get(url).content

# Parse HTML file.
parsed_html = BeautifulSoup(html_doc)

# Print parsed HTML.
print(parsed_html)

# Assign rows to table.
table = parsed_html.find_all('tr')

# Print table.
print(table)
```

## Programming

```python
## Create a list of sublists (the values for each row in the table).
# Extract <th> from first row of table and remove whitespace. Results assigned to 'headings'.
headings = [th.text.strip() for th in table[0].find_all('th')]

# Extract <td> from table (excluding header row) and remove whitespace.
# Create sublist for each row using list comprehension. Final results assigned to 'rows'.
rows = [[td.text.strip() for td in row.find_all('td')] for row in table[1:]]

# Print lists.
print(headings)
print(rows)
```

```python
## Create a data frame containing the values of the list in a table using pandas.
df = pd.DataFrame(rows, columns=headings)
```
