# 7. Most Common Word and Its Count

Find the most commonly occurring word in `overview_lower` and how many times it occurs. Store the word in a variable named `most_common_word` and the number of occurrences in a variable named `most_common_count`.

```python
# What is the most commonly occurring word and how many times does it occur?
import collections
most_com = collections.Counter(c).most_common(1)
most_common_word = most_com[0][0]
most_common_count = most_com[0][1]

print(most_common_word, most_common_count)
```
