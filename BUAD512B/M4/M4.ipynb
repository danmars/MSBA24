# Module 4: Data Wrangling

Data wrangling is the process that transforms raw data into formats that can be more readily analyzed and used for business modeling. In this module, we begin a deeper investigation of the `pandas` package and its ability to help us perform the necessary tasks of data wrangling. All projects and business modeling efforts spend substantial time and effort on the various data wrangling activities, such as discovering, structuring, cleaning, enriching, and validating the data. The process is also naturally iterative. The goal is to create tidy data that can then be analyzed efficiently and effectively.


## Learning Objectives

* Identify problematic areas within datasets
* Determine mitigation strategies when encountering missing data
* Reshape and aggregate data
* Create tidy data
