# Problem 1: Working with `DataFrame`s

Be sure to complete these steps in the order outlined below and in their appropriate code cells. There is file named `META.csv` in the subfolder `data`. It contains the trading data for the company Meta.

1. Use `pandas` to read the file into a `DataFrame` named `meta`, being sure to use a `RangeIndex` and making sure any date columns come in as a true date object.
2. Convert all the column names to lower case characters. Replace any spaces between words with the underscore character. These changes should be reflected in the existing `meta` `DataFrame`.
3. Change the index for `meta` to be the `date` column.
4. Using your newly created index (the date), add two columns to `meta`: (1) the name of the day as `day_name` (e.g., Monday, Tuesday, etc.) and (2) the numerical day of the week as `day_number` (e.g., 0, 1, 2, etc.).

```python
# Use `pandas` to read the file into a `DataFrame` named `meta`, 
# being sure to use a `RangeIndex` and making sure any date 
# columns come in as a true date object.

# import pandas using the alias pd
import pandas as pd

meta = pd.read_csv('./data/META.csv')
meta['Date'] = pd.to_datetime(meta['Date'])

meta.info()
```

```python
# This is a test cell
# If **NO** message is printed, it means that the tests passed
# One basic test to see if your code works
assert_in('meta', dir(), msg='You did not name your variable `meta` as instructed')
assert_equal(type(meta), pd.DataFrame, msg='You variable `meta` is not a DataFrame')
assert_equal(type(meta.index), pd.core.indexes.range.RangeIndex, msg='You do not have a RangeIndex for the DataFrame')
```

<hr>

```python
# Convert the column names to have all lower case characters.
# Replace any spaces between words with the underscore character. 
# These changes should be reflected in the existing `meta` `DataFrame`.
meta.columns = [col.lower().replace(" ", "_") for col in meta.columns]
meta.info()
```

<hr>

```python
# Change the index for `meta` to be the `date` column.
meta = meta.set_index('date')
meta.info()
```

<hr>

```python
# Using your newly created index, add two columns to `meta`: 
# (1) the name of the day as `day_name` and 
# (2) the numerical day of the week as `day_number`.
meta = meta.assign(day_name=lambda x: x.index.day_name(), day_number=lambda x: x.index.dayofweek)
meta
```

# Problem 2: Working with Data

Be sure to complete these steps in the order outlined below and in their appropriate code cells. There is file named `customers.csv` in the subfolder `data`. It contains information on the customers of one of your clients.
1. Read the file into a `DataFrame` named `cust`.
2. Find all the rows where the customer id is repeated; that is, mark all duplicates as `True`. Store the results in a `DataFrame` named `the_dupes`.
3. Which customer id is the most frequently occurring? Find it programmatically and store it in a variable named `cust_with_most_occurrences`.
4. Drop **all** the rows from `cust` that have a repeated customer id, being sure to store the results back into the variable `cust`.
5. Programmatically find the columns that have missing data and store their names in a list named `col_with_missing_data`. 
6. Replace any missing values with 'unknown' if the variable is categorical and the average of the variable if the column is quantitative. Be sure to store the results back into the variable `cust`.

```python
# Read the file into a `DataFrame` named `cust`
cust = pd.read_csv('./data/customers.csv')
cust.info()
cust.shape
```

```python
# This is a test cell
# If **NO** message is printed, it means that the tests passed
# One basic test to see if your code works
assert_equal(type(cust), pd.DataFrame, msg='You do not have a DataFrame named `cust`')
```

<hr>

```python
# Find all the rows where the customer id is repeated. 
# Store the results in a `DataFrame` named `the_dupes`
the_dupes = cust[cust.duplicated(subset=['cust_id'])]
the_dupes.info()
the_dupes.shape
```

```python
# This is a test cell
# If **NO** message is printed, it means that the tests passed
# One basic test to see if your code works
assert_equal(type(the_dupes), pd.DataFrame, msg='You do not have a DataFrame named the_dupes')
```

<hr>

```python
# Which customer id is the most frequently occurring? Find it programmatically 
# and store it in a variable named `cust_with_most_occurrences`.
cust_with_most_occurrences = the_dupes['cust_id'].mode()[0]
cust_with_most_occurrences
```

```python
# This is a test cell
# If **NO** message is printed, it means that the tests passed
# Did you correctly name the variable?
assert_in('cust_with_most_occurrences', dir(), msg="You did not name your variable `cust_with_most_occurrences` as instructed.")
```

<hr>

```python
# Drop all the rows from `cust` that have a repeated customer id, 
# being sure to store the results back into the variable `cust`
cust = cust.drop_duplicates(subset='cust_id')
cust.shape
```

<hr>

```python
# Programmatically find the columns that have missing data and 
# store their names in a list named `col_with_missing_data`

col_with_missing_data = cust.columns[cust.isna().any()].tolist()
col_with_missing_data
```

```python
# This is a test cell
# If **NO** message is printed, it means that the tests passed
# One basic test to see if your code works
assert_equal(type(col_with_missing_data), list, msg='You did not create a list')
```

```python
# Replace any missing values with 'unknown' if the variable is categorical 
# and the average of the variable if the column is quantitative.
# Be sure to store the results back into the variable `cust`.

cust['gender'].dtype

def replace_missing(data):
    for col in data:
        if data[col].isna().any() == True:
            if data[col].dtype == 'O':
                data[col].fillna('unknown', inplace=True)
            elif data[col].dtype == 'int64':
                data[col].fillna(col.mean())
        else:
            continue
    return data            

replace_missing(cust)
cust.info()
```
