## Renaming Columns

There are various reasons you may want to rename the columns of a `DataFrame`. Perhaps the column label is too vague or it is not descriptive enough for your taste. Maybe you want to replace any spaces found in the column name with underscores. In any event, renaming the columns of a `DataFrame` can be accomplished with the `.rename()` function. 

Let's try renaming columns from the long dataset that we saw earlier. 

```python
import pandas as pd
```

```python
# Read the data into a variable named long_df
long_df = pd.read_excel('./data/data_formats.xlsx', 'long')

# Look at the .info()
long_df.info()
```

```python
# Look at the dataframe itself
long_df
```

The column labeled `value` is actually the temperature in Fahrenheit. We want to rename the `value` column `temp_F`. The column labeled `variable` is somewhat ambiguous. Perhaps a better name for it would be `measurement`, so let's rename it as such.

For the argument `columns`, you pass a dictionary with the original column name as the key and the newly desired column name as the value. We will also perform the renaming in place.

```python
# Rename the column 'variable' to 'measurement'
# and the column 'value' to 'temp_F'
long_df.rename(columns={'variable': 'measurement', 'value': 'temp_F'},
               inplace=True)
long_df.info()
```

### Transforming All Column Names

Perhaps you prefer having all your column names as uppercase (e.g., you come from a database background and that's how you have always created your SQL tables). This transformation can be easily accomplished with the `.rename()` function as shown below.

```python
# Convert all column names to uppercase
up_long_df = long_df.rename(str.upper, axis='columns')
up_long_df.info()
```

### Adding a Column

Perhaps we not only want the temperature in the Fahrenheit scale but also the Celsius scale. We can add a new column named `temp_C` based off of the column `temp_F`. Recall that the conversion is:

$$
(F^{\circ} - 32) \times \frac{5}{9}
$$

We have seen how to create new columns from an existing column when we are simply scaling it with a single number (go back to M3 if you have forgotten how to do this). Here, we demonstrate another way to create a new column by using the `.assign()` method. The `.assign()` method is more flexible. For example, you can create multiple columns within the same assign. See the [documentation](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.assign.html) for more details.

Here we are only adding one column. We might as well do it both ways to see if the results are the same.

```python
# The way you have seen before in M3
# First make a copy of long_df
one_way = long_df.copy()
one_way.info()
```

```python
# Add the column to one_way
one_way['temp_C'] = (one_way['temp_F'] - 32) * 5/9
one_way.info()
```

```python
# Look at one_way
one_way
```

```python
# Look at long_df 
long_df.info()
```

```python
# Use the .assign() method which returns a new DataFrame
another_way = long_df.assign(temp_C = (long_df.temp_F - 32) * 5/9)
another_way.info()
```

```python
# Look at another_way
another_way
```

## Type Conversion

There are many instances when you will have to convert the data types associated with the columns of a `DataFrame`. This process is often referred to as **casting** a column from one data type to another. For example, you may have a `.csv` file that contains a column that is date, but when you read it in, the data type does not fully recognize it as a timestamp or date. In other instances, you may have a categorical variable that you want to explicitly define that way instead of the default of `object`, or perhaps, you want to convert whole numbers to decimals. Let's examine a few of these with our daily temperature data currently stored in the `DataFrame` named `another_way`. 

```python
# Look at .info on another_way
another_way.info()
```

Our column labeled `date` looks like it is already a date -- good. Both of the temperature columns also appear to be fine as floating point decimal numbers. The `measurement` column is an `object`, which generally means it contains strings. Let's try to get the summary statistics for all the columns to see if we should convert the data type of `measurement` to a categorical one.

```python
# Get summary statistics
another_way.describe(include='all', datetime_is_numeric=True)
```

We see that the `measurement` column only contains two unique values. This should not be surprising: we had `min_temp` and `max_temp`, the low and high temperatures recorded for the day. Let's cast it to a `category` data type.

```python
# Casting `measurement` from `object` to `category`
temps_df = another_way.assign(measurement=another_way.measurement.astype('category'))
temps_df.info()
```

Just for fun, let's create two new columns that hold the whole number representations of the two temperature columns. You could think of this as, "Do I really care if the temperature is 45 degrees versus 45.6 degrees as I prepare to leave my house?" To create these new columns, we will use the `.assign()` function. 

```python
# Create two new columns in temps_df
# One for whole number temp_F and one for whole number temp_C
temps_df = temps_df.assign(
    temp_F_whole=temps_df.temp_F.astype('int'),
    temp_C_whole=temps_df.temp_C.astype('int'))

# See if it worked
temps_df.info()
```

```python
# Look at the new temps_df
temps_df
```

## Reordering, Reindexing, and Sorting Data

You will often want to sort your data by the values of one or more columns. For example, which days had the highest low temperatures (e.g, biggest `min_temp`) in Williamsburg for the month of December 2022. One approach to finding this information is to sort our `temps_df` by one of the temperature columns in descending order and look at the `head()`. Let's try it.

```python
# Looking at min_temp, sorting by temp_C_whole in descending order
temps_df[temps_df.measurement == 'min_temp'].sort_values(by='temp_C_whole', ascending=False).head(10)
```

Notice that in the results above that both December 31st and December 6th have the same whole Celsius temperature values of six degrees, but that December 31st is listed first. If you look at the more precise `temp_C` column, then these rows are out of order. We see something similar when looking at the last four rows. One obvious way to approach sorting is to use the more precise column rather than the column `temp_C_whole` like we did.

What if we are only interested in the whole number temperatures, but wanted them sorted secondarily by the date? This is easily accomplished by sending a list to both the `by` and the `ascending` arguments.

```python
# Looking at min_temp, sorting by temp_C_whole in descending order
# and secondary sorting by date in ascending order
temps_df[temps_df.measurement == 'min_temp'].sort_values(
    by=['temp_C_whole', 'date'], ascending=[False, True]).head(10)
```

There is another way to look at a subset of sorted values using `pandas`; we can use the functions `nlargest()` or `nsmallest()`. Let's try it.

```python
# Use nlargest() to find top 10 based on temp_C_whole
temps_df[temps_df.measurement == 'min_temp'].nlargest(n=10, columns='temp_C_whole')
```

```python
# Use nsmallest() to find bottom 10 based on temp_C_whole
temps_df[temps_df.measurement == 'min_temp'].nsmallest(n=10, columns='temp_C_whole')
```

### Useful DatetimeIndex

Let's now look at some more time series data to see another useful advantage of using a `DatetimeIndex`. We will be using two different datasets: `sp500_2022.csv` and `bitcoin_2022.csv`. These each contain the opening, high, low, closing, and adjusted closing price, along with the total volume for each trading day for the year 2022.

```python
# Get the S&P 500 data and look at it
sp500 = pd.read_csv('./data/sp500_2022.csv')
sp500.info()
```

```python
# Take a peek
sp500
```

Notice that all the data types came in as `object`, but it appears that we have a real date for the `Date` column and numbers for the other columns. Why are the numbers coming in as `object`? I suspect it is the thousands separator (commas). Let's try to re-read the data in. This time let's try to make the `Date` column the index, drop the `Adj Close` column, and see if we can force the other columns to be numbers.

```python
sp500 = pd.read_csv('./data/sp500_2022.csv',
                   index_col = 'Date', parse_dates = True,
                   thousands = ',').drop(columns=['Adj Close'])

sp500.info()
```

```python
# Take a peek
sp500
```

```python
# Use same method for the bitcoin data
# Get the bitcoin data and look at it
bitcoin = pd.read_csv('./data/bitcoin_2022.csv',
                     index_col = 'Date', parse_dates = True,
                     thousands = ',').drop(columns=['Adj Close'])
bitcoin.info()
```

```python
# Take a peek
bitcoin
```

### Finding the Name of the Day
 
When we have a `DatetimeIndex`, we can easily find the name of the day (e.g., Monday, Tuesday, etc.) that the date refers to. This ability is quite useful for time series analysis. Let's find the day of the week for the S&P 500 data and add it as a new column to `sp500`. We will also do the same for the `bitcoin` data.

```python
# Add the column 'day_name' and put back into sp500
sp500 = sp500.assign(day_name=lambda x: x.index.day_name())

# See the results
sp500
```

```python
# Add the column 'day_name' and put back into bitcoin
bitcoin = bitcoin.assign(day_name=lambda x: x.index.day_name())

# See the results
bitcoin
```

## Transposing DataFrames

Most of the time you will be manipulating data in either the wide or long format. However, there will be times that you may want the data structured differently. `pandas` is quite flexible and allows many different manipulations to get the data in the format you desire. One manipulation you may see is **transposing** the data. When you transpose a dataset, the rows become the columns and the columns become the rows. You probably have experience with this concept from either using MS Excel or a linear algebra course you have taken. While it may not be immediately apparent why formatting the data in its transposed form is beneficial, it can be as simple as making better use of our display area. Some visualizations may be easier to build after you transpose the data. Let's try it. 

```python
# Look at temps_df
temps_df
```

```python
# Transpose our temps_df DataFrame
temps_df.T
```

Notice that the column headers are numbers. Those came from the `RangeIndex`. For this dataset, it would make sense to use the date as the column headers. Let's try transposing again, but first set the index to the `date` column.

```python
# Set index to `date` and then transpose
temps_df.set_index('date').T
```

```python
# It does not "stick", but rather returns a new DataFrame
temps_df
```

## Pivoting DataFrames

When we **pivot** a dataset, we are going from long format to wide format. You need to specify which column currently holds the values and which column contains what will become the column names in the wide format. You can also optionally specify a new index. Let's try pivoting our `temps_df`.

```python
pivoted_df = temps_df.pivot(index='date', columns='measurement', values='temp_F')
pivoted_df
```

```python
# Call .describe()
pivoted_df.describe()
```

We can provide multiple columns to the `values` argument as a list. Let's include `temp_C`.

```python
pivoted_df = temps_df.pivot(index='date', columns='measurement',
                            values=['temp_F','temp_C'])
pivoted_df
```

Well, that looks different than what we have seen before. What happened? We created a **hierarchical index**. If we want to select `max_temp`, we first have to specify whether we want the temperatures in Fahrenheit or Celsius. Let's look at the `.info()` and then try to get the `max_temp` for `temp_F`.

```python
# What does .info() show now?
pivoted_df.info()
```

```python
# First ask for `temp_F` and then `max_temp`
pivoted_df['temp_F']['max_temp']
```

We have thus far only been using a single index. However, we can create an index from any number of columns with the function `set_index()`. When we use more than one column to create an index, we create what is called a `MultiIndex`. The outermost level of the `MultiIndex` corresponds to the first element in the list when we call `set_index()`. Let's try creating a new `DataFrame` with `MultiIndex` of `date` and `measurement`.

```python
# Create new DataFrame with MultiIndex
multi_index_temps = temps_df.set_index(['date', 'measurement'])
multi_index_temps.info()
```

```python
# Pull out just the index 
multi_index_temps.index
```

```python
# What does the DataFrame look like?
multi_index_temps
```

```python
# Look at only December 15th
multi_index_temps.loc['2022-12-15']
```

### Unstacking

The `pivot()` method expects the data to only have one column set as the index. If you have a multi-level index, then you must **unstack** the data to get back to a single column as the index. To perform this task, you use the `unstack()` function. By default the `unstack()` method will move the innermost level of the index to the columns. If you want to unstack in a different order, you need to pass the index of the desired level to `unstack()`, where 0 is the leftmost and -1 is the rightmost. You can also pass the name of the level you want to unstack, assuming it has one.

```python
# Try unstacking using default arguments
multi_index_temps.unstack()
```

```python
# Try unstacking using the index of -1
# Should result in same as above
multi_index_temps.unstack(-1)
```

```python
# Try unstacking using name 'measurement'
# Should result in same as above
multi_index_temps.unstack('measurement')
```

```python
# Try unstacking using name 'date'
# Result should be DIFFERENT
multi_index_temps.unstack('date')
```

## Melting DataFrames

When we want to go from wide format to long format, we **melt** the data. In essence, melting undoes a pivot. Melting requires that we specify a few things. Using the `id_vars` argument, we need to specify which column(s) uniquely identify a row in the wide format data. The `value_vars` argument tells it which column(s) contain(s) the variable(s). Optionally, we can specify how to name the column containing the variable names in the long format data using `var_name` and the name for the column containing their values using `value_name`. If you do not specify these arguments, they will be `variable` and `value`, respectively.

Let's read in the temperature data as wide format and see if we can melt it to a long format. Recall that data is in the file `./data/data_formats.xlsx`.

```python
# Read in the wide format
wide = pd.read_excel('./data/data_formats.xlsx', 'wide')
wide.info()
```

```python
# Look at it
wide
```

```python
# Try melting
wide.melt(id_vars=['date'])
```

```python
# Now let's try naming the columns something
wide.melt(id_vars=['date'], var_name='measurement', value_name='temp_F')
```

### Stacking

Similar to the way pivoting has an alternative with `unstack()`, there is an alternative for melting with the `stack()` method. The `stack()` method will pivot the columns into the innermost level of the index, resulting in a `MultiIndex`. Therefore, we should carefully inspect our index before using the `stack()` method. Let's see the method in action.

```python
# set the index of wide to be the `date` column
wide.set_index('date', inplace=True)
wide
```

```python
# Now try to stack the data
# This will result in a `Series` based on the structure of our data
stacked_series = wide.stack()
stacked_series
```

```python
# Turn the Series into a DataFrame
stacked_df = stacked_series.to_frame('values')
stacked_df
```

```python
# Look at .info()
stacked_df.info()
```

```python
# And the index by itself
stacked_df.index
```
