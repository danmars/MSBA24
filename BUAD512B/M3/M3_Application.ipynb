# 2. Liftgate Service

When customers request a quote to move their freight, they need to answer whether or not a liftgate service is needed at either the origin, the destination, or both. In the request data, this information is captured as a `1` if the service is required and a `0` if it is not. (The image below shows a typical liftgate on a trailer.) You want to determine how many requests:

1. Require a liftgate service at the origin, regardless if one is required at the destination
2. Require a liftgate service at the destination, regardless if one is required at the origin
3. Require a liftgate service at **both** the origin and destination

For the first question, store your answer in a variable named `needed_at_origin`. For the second question, create a variable named `needed_at_destination` for your answer. The answer to the third question should be stored in a variable named `needed_at_both`. 

![An example of a liftgate](images/liftgate.jpg)

```python
request_data = pd.read_excel('./data/pm_data.xlsx', sheet_name=2)

origin_liftgate = (request_data["origin_liftgate_service"] == 1)
destination_liftgate = (request_data["destination_liftgate_service"] == 1)
both_liftgate = (request_data["origin_liftgate_service"] == 1) & (request_data["destination_liftgate_service"] == 1)

needed_at_origin = origin_liftgate.sum()
needed_at_destination = destination_liftgate.sum()
needed_at_both = both_liftgate.sum()

print(needed_at_origin, needed_at_destination, needed_at_both)

```
