# Introduction to the `csv` Module

We have seen how to read in data from comma-separated value formatted files by first opening the file with the standard `open()` command and then either using `read()` or `readlines()`. Unsurprisingly, there is a module called `csv` that takes care of some of the tedious overhead when dealing with these files. You will undoubtedly encounter code that uses the `csv` module. Therefore, our primary purpose is to be aware of the module and its abilities. You will soon see that we have another, more robust package to help us with reading and manipulating `.csv` files. 

## Using the `csv` Module

We first import the `csv` module with our standard import statement. We still need to open a file and we will use best practices by doing so inside the context of a `with` statement. As you may recall, the standard `open()` command gives us back a `TextIOWrapper` object. When we use the csv module we create a `_csv.reader` object and use it. 

Let's look at an example. First, we'll just open the `states.csv` file and pass the file handle to the `csv.reader()` and print out its return type. Then, we will open the file again and simply print out the contents of the file.

One thing to note is that we will be using the argument `newline=''`. As the documentation states: "If `newline=''` is not specified, newlines embedded inside quoted fields will not be interpreted correctly ... It should always be safe to specify `newline=''`, since the csv module does its own (universal) newline handling."

```python
# import the csv module
import csv
```

```python
# Open states.csv and create a reader, print out its type
with open('./data/states.csv', 'r', newline='') as f:
    reader = csv.reader(f)
    print(type(reader))
```

```python
# Open the states.csv file
with open('./data/states.csv', 'r', newline='') as f:
    # create a csv.reader object
    reader = csv.reader(f)
    # next(reader) returns the next row
    # In our case it is the header
    print(next(reader))
```

```python
# Open the states.csv file
with open('./data/states.csv', 'r', newline='') as f:
    # create a csv.reader object
    reader = csv.reader(f)
    # print the header row
    print('First Row = Header Row')
    print(next(reader))
    print()
    
    # now loop over the reader object, printing each line
    print('Data Rows Follow:')
    for row in reader:
        print(row)
```

### Writing Files

We can also easily write data out to files using the `csv` module by creating and using a `writer` object. Again, we should follow best practices and write the file in the context of a `with` statement. 

For this example, let's read the header row from `states.csv` and store it as a variable. Then, let's store every even row of data into a list. After that, we'll print the header row and the even rows of data into a new file.

```python
# Create an empty list to hold even rows
even_rows = []

# Open the states.csv file
with open('./data/states.csv', 'r', newline='') as f:
    # create a csv.reader object
    reader = csv.reader(f)
    # store the header row in variable named header
    header = next(reader)
    
    # now loop over the reader object getting even rows
    # initialize an index to 0
    i = 0
    for row in reader:
        # increment the index each time
        i += 1
        if i % 2 == 0:
            even_rows.append(row)
        else:
            continue
            
# What does header look like?
print(f'Header looks like: {header}')
print('\nand here is our data:')
print(even_rows)
```

Looks like that worked. Now, let's write this data into a new file. The function `writerow()` writes a single row. If we have an *iterable* object, such as our list, and want to write each element in the file, then we can use the `writerows()` function. Below, we use `writerow()` to write the header and `writerows()` to write all of our even rows of data.

```python
# Open a file called even_rows.csv for writing
with open('./data/even_rows.csv', 'w', newline='') as f:
    # Create a writer object using file handle
    writer = csv.writer(f)
    
    # Write the header row
    writer.writerow(header)
    
    # Write the data
    writer.writerows(even_rows)
```
