# Problem 1: Counting Letters

- Implement a function called `count_characters` that takes in two arguments: a string representing the `full_text` and a second string representing the `characters` to be counted.
- The function should return a dictionary containing the character as the key and the number of occurrences found in the `full_text` as the value.
- **NOTE:** You should return the number of characters for only the characters found in string `characters` argument. Additionally, you should count uppercase and lowercase letters as the <ins>**same**</ins> letter.

```python
def count_characters(full_text, characters):
    """
    A function that takes in some text and counts the specified characters.
    
    Parameters
    ----------
    full_text: The text to search
    characters: The characters to look for
    
    Returns
    -------
    count_dict: A dictionary containing the character as the key and the 
                 count as the value
    """
    
    # Initialize the dictionary to return
    count_dict = {}
    for char in full_text.lower():
        if char in characters.lower():
            count_dict[char] = count_dict.get(char, 0) + 1    
    return count_dict

count_characters('Hello Earnest!', 'aeiou')
```

# Problem 2: Find Top 3 Rows

Implement a function called `find_top_3_rows` that takes in a `pandas` `DataFrame` (i.e., the argument named `the_data`), and a specified column name from that `DataFrame` (argument named `the_col`), and returns a `DataFrame` called `ret_df` that contains the top 3 rows with the **largest** values from the passed in column, `the_col`.

```python
def find_top_3_rows(the_data, the_col):
    """
    A function that receives a DataFrame and the name of a column in that DataFrame.
    It should return a new DataFrame with the top 3 rows with the largest values
    from the column specified.
    
    Parameters:
    -----------
    the_data: a DataFrame
    
    the_col: the name of the column in theData that you are looking for largest values of
    
    Returns:
    --------
    ret_df: a DataFrame containing the top three rows of the passed in theData based on 
           the column name passed in, theCol
    """
    sort_df = the_data.sort_values(by = the_col, ascending = False)
    ret_df = sort_df.head(3)
    
    return ret_df
```

# Problem 3: Reading File and Returning a `list`

There is file named `restaurant.csv` in the subfolder `data`. Create a function called `get_rows_as_list` that takes in the name of the subdirectory (i.e., the argument `sub_dir`) and the name of the file (i.e., the argument named `the_file`), and returns a `list` called `ret_list` that returns each row of the file as an element of the list.

```python
def get_rows_as_list(sub_dir, the_file):
    '''
    A function that receives the name of a subfolder/subdirectory and the name 
    of a file in that subfolder. The function returns a list where each element
    of the list is a row in the file.
    
    Parameters:
    -----------
    sub_dir: the name of the subfolder/subdirectory
    
    the_file: the name of the file that you want to read
    
    Returns:
    --------
    ret_list: a list containing each row of the file as an element of the list
    '''
    ret_list = []
    file_path = ('./' + sub_dir + '/' + the_file)
    with open(file_path, 'r') as file:
        ret_list = file.readlines()    
    return ret_list
```

# Problem 4: Reading File with the `csv` Module

- Import the appropriate module / package.
- There is file named `restaurant.csv` in the subfolder `data`. 
- Read the first row of data from the `restaurant.csv` file into a variable called `my_header`.
- Read every **third** row of data into a list named `every_third_row`.
- Write both the header and `every_third_row` to a new file named `out_file.csv` in the subfolder `data`.

```python
# import the appropriate module/package
import csv

every_third_row = []

with open('./data/restaurant.csv', 'r', newline='') as f:
    reader = csv.reader(f)
    my_header = next(reader)
    i = 0
    for row in reader:
        i += 1
        if i % 3 == 0:
            every_third_row.append(row)
        else:
            continue
            
with open('./data/out_file.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(my_header)
    writer.writerows(every_third_row)
            
    
```
