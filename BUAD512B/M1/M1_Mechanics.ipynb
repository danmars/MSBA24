## Problem 6. Strings

In the code cell below is a string named `spooky_story`. You will be using that string for this problem. Be sure to run that code cell before trying to tackle the tasks below.

```python
# Run this code cell
spooky_story = 'I always thought my cat had a staring problem - \
she always seemed fixated on my face. Until one day, when I realized \
that she was always looking just behind me.'

print(spooky_story)
```

```python
# 1. How many characters are in `spooky_story`? Store the result in a the variable named `num_chars`.
num_chars = len(spooky_story)
print(num_chars)
```

```python
## DO NOT DELETE
# This is a test cell
# If **NO** message is printed, it means that the tests passed
# Did you correctly name the variable `num_chars`?
assert_in("num_chars", dir(), msg="You did not name your variable `num_chars` as instructed.")
```

<hr>

```python
# 2. Break `spooky_story` into a list using space as the delimiter. Name the list `words`.
words = spooky_story.split()
print(words)
```

<hr>

```python
# 3. What is the fifth word `words`? Store the result in a variable named `fifth_word`.
fifth_word = words[4]
print(fifth_word)
```
