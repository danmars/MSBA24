### Creating DataFrames from RDDs

```python
from pyspark.sql import SparkSession
from pyspark import SparkContext
from pyspark.sql.types import *
sc = SparkContext.getOrCreate()
spark = SparkSession.builder.getOrCreate()
```

```python
def mapper(line):
    fields = line.split(',')
    return (str(fields[1]), str(fields[2]), int(fields[8]))
```

```python
lines = sc.textFile("SalesRecords2.csv")
salesRDD = lines.map(mapper)
```

```python
mySchema = StructType([StructField("Country", StringType (), True), \
                       StructField("Item", StringType(), True), \
                       StructField("Quantity", IntegerType(), True)])
```

```python
salesDF = spark.createDataFrame(salesRDD, mySchema)
salesDF.show(5)
```

```python
salesDF.where("Item in ('Fruits', 'Vegetables')").orderBy("Quantity").show(10)
```

```python
#Creates a temporary view using the DataFrame
salesDF.createOrReplaceTempView("salesDFview")
spark.sql("SELECT * FROM salesDFview WHERE Item in ('Fruits', 'Vegetables') order by Quantity").show(10)
```
