# Spark SQL and Data Frames

Data Frames
- contain row objects
- have schema
- read and write to JSON, Hive, parquet, etc.



### Create DataFrame via User Input

```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
```

```python
myCustList = [["Mary","Luthans"],["Joyce","Lari"],["Kathi","Burge"]]
myCustDF = spark.createDataFrame(myCustList)
myCustDF.show()
```

```python
newCustDF = myCustDF.select("_1", "_2").withColumnRenamed("_1","firstName")\
                                        .withColumnRenamed("_2","lastName")
newCustDF. show() 
```

### Loading Files into DataFrames

```python
customersDF = spark.read.json("customers.json")
customersDF.show()
```

```python
customersDF.printSchema()
```

```python
customersDF = spark.read.format("csv").option("header", "true").load("customers.csv")
customersDF.show() 
```
