## Analyzing DataFrames

```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
```

```python
customersDF = spark.read.format("csv").option("header", "true").load("customers.csv")
```

```python
ageCol = customersDF.select("Age")
ageCol = customersDF.select(customersDF["Age"])
ageCol = customersDF.select(customersDF.Age) 
```

Examples of Column Operations in Spark
- Arithmetic Operations: +, -, %, *
- Comparative Operations: >, <, &&, ||, ===
- String Functions: like, contains, substr
- Data Testing Operations: isNull, isNotNull, NaN
- Sorting Functions: sort, orderBy, asc, desc


```python
ageDF = customersDF.select("FirstName", "LastName", "Age").where(customersDF.Age.isNotNull())
ageDF.show()
```

```python
customersDF.groupBy("Address").count().show()
```

### Joining DataFrames

```python
zipCodeDF = spark.read.format("csv").option("header", "true").load("zipcodes.csv")
zipCodeDF.show()
customersDF.show()
```

```python
customersDF.join(zipCodeDF, "City").show()
```

```python
customersDF.join(zipCodeDF, "City", "left_outer").show()
```
