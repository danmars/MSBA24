```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
```

```python
# Read the data from the parquet file, make sure to revise the bucket name as needed
trips_df = spark.read.parquet("gs://adurukansonmez-5132/spark/yellow_tripdata_2023-01.parquet")
```

```python
# Print the first 10 observations from trips_df
trips_df.show(10)
```

```python
print("There are {} rows in the trips_df DataFrame.".format(trips_df.count()))
```

```python
trips_df.printSchema()
```

```python
# Select passenger_count, trip_distance, tip_amount and save it to a new dataframe:
trips_df_sub = trips_df.select('passenger_count', 'trip_distance', 'tip_amount')
trips_df_sub.show(10)
```

```python
# Print the number of records in trips_df_sub
trips_df_sub.count()
```

```python
# Remove duplicate entries from trips_df_sub
trips_df_sub_nodup = trips_df_sub.dropDuplicates()
# Print the number of records in trips_df_sub_nodup
trips_df_sub_nodup.count()
```

```python
# Delete the records that have missing value(s)
trips_df = trips_df.na.drop()
trips_df.count()
```

```python
# Filter the records where there passenger count is 1
trips_df_1pass = trips_df.filter(trips_df.passenger_count == 1)
trips_df_1pass.count()
```

```python
# Count the number of records where the passenger count is higher than 2
trips_df.filter(trips_df.passenger_count > 2).count()
```

```python
# Summary statistics for the number of passengers per trip
trips_df.describe('passenger_count').show()
```

```python
# Summary statistics for all variables
trips_df.describe().show()
```

```python
# Display the average value for trip distance column
trips_df.agg({'trip_distance': 'mean'}).show()
```

```python
# Number of trips by passenger count
trips_df.groupby('passenger_count').count().show()
```

```python
# Number of trips by passenger count, sorted in ascending order of passenger count
trips_df.groupby('passenger_count').count().orderBy(trips_df.passenger_count.asc()).show()
```

```python
# Number of trips by passenger count, sorted in descending order of count
trips_df.groupby('passenger_count').count().orderBy('count', ascending=False).show()
```

```python
# Find the average trip distance for trips by number of passengers
trips_df.groupby('passenger_count').agg({'trip_distance': 'mean'}).show()
```

```python
# Find the average tip amount for trips by number of passengers
trips_df.groupby('passenger_count').agg({'tip_amount': 'mean'}).show()
```

```python
# Register your dataframe as a table
trips_df.createOrReplaceTempView('trips_table')
# Retrieve all records where the passenger count is 8 and the trip distance is between 0.1 and 5 
trips2 = spark.sql("SELECT passenger_count, trip_distance \
                   FROM trips_table \
                   WHERE passenger_count == 8 \
                   AND trip_distance BETWEEN 0.1 AND 5")
# trips2 is the new dataframe
trips2.show()
```
