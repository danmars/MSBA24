```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
```

```python
friendsDF = spark.read.format("csv").option("header", "false").load("friends.csv")
friendsDF.show(10)
```

```python
friendsDF  = friendsDF.select("_c0","_c1","_c2","_c3").withColumnRenamed("_c0","ID").withColumnRenamed("_c1","Name")\
                                                      .withColumnRenamed("_c2","Age").withColumnRenamed("_c3","numFriends")
friendsDF.show(10)
```

```python
friendsDF.createOrReplaceTempView("friendsDFview")
teenagers = spark.sql("SELECT * FROM friendsDFview WHERE age >= 13 AND age <= 19")
teenagers.show(10)
```

```python
spark.sql("SELECT age, COUNT(*) FROM friendsDFview GROUP BY age ORDER BY 2 DESC").show()
```
